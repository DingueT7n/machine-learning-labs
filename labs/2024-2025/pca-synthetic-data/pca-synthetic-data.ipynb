### 1. Generate a 3D dataset following a normal distribution and visualize it



```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Generate 500 vectors in 3D space following a normal distribution
rndn3d = np.random.randn(500, 3)

# Visualize the 3D point cloud
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(rndn3d[:, 0], rndn3d[:, 1], rndn3d[:, 2])
plt.title("Initial 3D Data")
plt.show()
```

### 2. Apply PCA to the 3D data and analyze the results



```python
from sklearn.decomposition import PCA

# Apply PCA
pca = PCA(n_components=3)
pca.fit(rndn3d)

# Print explained variance ratio and principal components
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("Principal Components:")
print(pca.components_)
```

### 3. Generate a new dataset and compare PCA results

```python
# Generate another dataset
rndn3d_new = np.random.randn(500, 3)

# Apply PCA to the new dataset
pca_new = PCA(n_components=3)
pca_new.fit(rndn3d_new)

# Compare explained variance ratio and principal components
print("Explained Variance Ratio (New Data):")
print(pca_new.explained_variance_ratio_)

print("Principal Components (New Data):")
print(pca_new.components_)
```

### 4. Apply deformation and rotation to the original dataset

```python
# Define a deformation matrix
deformation_matrix = np.array([[2, 0, 0], 
                                [0, 1, 0], 
                                [0, 0, 0.5]])

# Apply deformation
rndef = rndn3d @ deformation_matrix.T

# Visualize the deformed data
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(rndef[:, 0], rndef[:, 1], rndef[:, 2])
plt.title("Deformed 3D Data")
plt.show()
```

```python
# Apply PCA to the deformed data
pca_def = PCA(n_components=3)
pca_def.fit(rndef)

# Print explained variance ratio and principal components
print("Explained Variance Ratio (Deformed Data):")
print(pca_def.explained_variance_ratio_)

print("Principal Components (Deformed Data):")
print(pca_def.components_)
```

### 5. Apply the same transformation to a new dataset and compare results



```python
# Generate another dataset
rndn3d_new2 = np.random.randn(500, 3)

# Apply the same deformation
rndef_new = rndn3d_new2 @ deformation_matrix.T

# Apply PCA to the new deformed dataset
pca_def_new = PCA(n_components=3)
pca_def_new.fit(rndef_new)

# Compare results
print("Explained Variance Ratio (New Deformed Data):")
print(pca_def_new.explained_variance_ratio_)

print("Principal Components (New Deformed Data):")
print(pca_def_new.components_)
```
